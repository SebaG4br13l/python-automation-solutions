```python
# 1. INSTALACIÓN DE LIBRERÍAS DE IA Y DATOS (Obligatorio en Colab)
!pip install transformers torch pandas openpyxl

import pandas as pd
from transformers import pipeline
from google.colab import files

def analizar_resenas_con_ia():
    print("🤖 [PM Bot] Inicializando el modelo de Inteligencia Artificial en la nube...")

    # Descargamos un modelo gratuito de análisis de sentimiento optimizado para español
    # Este proceso puede tardar unos segundos la primera vez porque descarga el cerebro de la IA
    try:
        clasificador = pipeline(
            "text-classification",
            model="pysentimiento/robertuito-sentiment-analysis"
        )
        print("✅ Modelo de IA cargado exitosamente en los servidores de Google.")
    except Exception as e:
        print(f"❌ Error al cargar el modelo de IA: {e}")
        return

    # 📋 BASE DE DATOS SIMULADA (Lo que un cliente real te daría en un Excel)
    print("\n📦 Cargando lote de reseñas de clientes del restaurante...")
    datos_resenas = [
        "La comida estuvo deliciosa y la atención del mesero fue excelente, vuelvo seguro.",
        "El servicio fue sumamente lento, tardaron una hora en traer el plato y llegó frío.",
        "Un lugar agradable, los precios están bien pero el ambiente es un poco ruidoso.",
        "Pésima experiencia. La mesa estaba sucia y la comida no tenía buen sabor.",
        "Me encantó el postre, aunque el precio del estacionamiento me pareció muy caro."
    ]

    resultados_analisis = []

    print("\n🧠 [IA en Acción] Analizando textos y extrayendo sentimientos...")
    for i, resena in enumerate(datos_resenas, 1):
        # La IA analiza el texto de la reseña
        prediccion = clasificador(resena)[0]

        # Traducimos las etiquetas del modelo para que el reporte sea comprensible
        etiqueta_original = prediccion['label']
        if etiqueta_original == 'POS':
            sentimiento = "Positivo"
        elif etiqueta_original == 'NEG':
            sentimiento = "Negativo"
        else:
            sentimiento = "Neutral"

        porcentaje_certeza = prediccion['score'] * 100

        print(f"  ↳ Reseña #{i}: Clasificada como [{sentimiento}] con {porcentaje_certeza:.1f}% de certeza.")

        # Guardamos la información estructurada
        resultados_analisis.append({
            "ID Reseña": f"RES-{i:03d}",
            "Comentario del Cliente": resena,
            "Análisis de Sentimiento (IA)": sentimiento,
            "Confianza del Modelo": f"{porcentaje_certeza:.1f}%"
        })

    # 📊 EXPORTACIÓN DE ENTREGABLE COMERCIAL
    df = pd.DataFrame(resultados_analisis)
    archivo_salida = "auditoria_sentimientos_ia.xlsx"

    # Guardamos en el disco de Colab
    df.to_excel(archivo_salida, index=False)
    print(f"\n✅ Reporte analítico generado: '{archivo_salida}'")
    print("📥 Iniciando descarga automática a tu equipo...")

    # Descarga directa a tu laptop
    files.download(archivo_salida)

# Ejecutar el hilo de procesamiento inteligente
analizar_resenas_con_ia()
```
