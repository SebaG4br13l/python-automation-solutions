```python
# 1. INSTALACIÓN DE LIBRERÍAS (Obligatorio en Colab cada vez que abres el cuaderno)
!pip install requests beautifulsoup4 pandas openpyxl

# 2. CÓDIGO DEL EXTRACTOR ADAPTADO
import requests
from bs4 import BeautifulSoup
import pandas as pd
from google.colab import files  # <-- Librería especial para Colab

def extraer_mercado_libre_colab():
    # URL de departamentos en venta
    url = "https://realstateFalseurl.com/"

    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
    }

    print("🤖 [PM Bot] Iniciando extracción desde la nube de Google...")
    response = requests.get(url, headers=headers)

    if response.status_code != 200:
        print(f"❌ Error de conexión: Código {response.status_code}")
        return

    soup = BeautifulSoup(response.text, 'html.parser')
    propiedades = []
    tarjetas = soup.find_all('div', class_='ui-search-result__wrapper')

    print(f"📦 Se encontraron {len(tarjetas)} propiedades en la página.")

    for tarjeta in tarjetas:
        try:
            titulo_tag = tarjeta.find('h2', class_='ui-search-item__title')
            titulo = titulo_tag.text.strip() if titulo_tag else "Sin título"

            precio_completo = tarjeta.find('span', class_='andes-money-amount__fraction')
            moneda_tag = tarjeta.find('span', class_='andes-money-amount__currency-symbol')
            precio = precio_completo.text.strip() if precio_completo else "0"
            moneda = moneda_tag.text.strip() if moneda_tag else "$"

            ubicacion_tag = tarjeta.find('span', class_='ui-search-item__location')
            ubicacion = ubicacion_tag.text.strip() if ubicacion_tag else "No especificada"

            link_tag = tarjeta.find('a', class_='ui-search-link')
            enlace = link_tag['href'] if link_tag else ""

            propiedades.append({
                "Título": titulo,
                "Precio": f"{moneda} {precio}",
                "Ubicación": ubicacion,
                "Enlace": enlace
            })
        except Exception as e:
            continue

    if propiedades:
        df = pd.DataFrame(propiedades)
        archivo_salida = "reporte_inmobiliario.xlsx"

        # Guardamos el archivo en la nube de Colab
        df.to_excel(archivo_salida, index=False)
        print(f"✅ Archivo generado en la nube. Iniciando descarga automática...")

        # DESCARGA AUTOMÁTICA A TU COMPUTADORA REAL
        files.download(archivo_salida)
    else:
        print("⚠️ No se pudo extraer información.")

# Ejecutar la función
extraer_mercado_libre_colab()
```
